Question 2 : Use Copilot to:

Find total sales per city

Top 5 cities by revenue

Ans:To use copilot to find total sales and total cities revenue are

    -- Total Sales per City

    SELECT
    City,
    SUM(PurchaseAmount) AS TotalSales
    FROM SalesData
    GROUP BY City
    ORDER BY TotalSales DESC;


------------------------------------------------------------

    -- Top 5 Cities by Revenue

    SELECT TOP 5
    City,
    SUM(PurchaseAmount) AS Revenue
    FROM SalesData
    GROUP BY City
    ORDER BY Revenue DESC;

Question 10 : Generate VBA code using Copilot for cleaning data (removing duplicates, filling null values,
format table).

Ans: To generate vba code using a copilot for cleaning data are

    Sub CleanData()

    Dim ws As Worksheet
    Set ws = ActiveSheet

    ' Remove duplicate rows
    ws.Range("A1").CurrentRegion.RemoveDuplicates _
    Columns:=Array(1, 2, 3, 4, 5, 6), Header:=xlYes

    ' Fill blank cells with N/A
    Dim cell As Range

    For Each cell In ws.Range("A2:Z1000")
        If IsEmpty(cell.Value) Then
            cell.Value = "N/A"
        End If
    Next cell

    ' Format table
    With ws.Range("A1").CurrentRegion

        ' Bold headers
        .Rows(1).Font.Bold = True

        ' Auto fit columns
        .Columns.AutoFit

        ' Add borders
        .Borders.LineStyle = xlContinuous

    End With

    MsgBox "Data Cleaning Completed Successfully!"

    End Sub
